### Exercise 10 — Multiple Subplots

Create a `2 × 2` figure containing:

1. Line chart
2. Bar chart
3. Scatter plot
4. Histogram

Use `plt.subplots()`.

Give each subplot a title.

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 25]

x_scatter = [1, 2, 3, 4, 5]
y_scatter = [2, 5, 4, 8, 7]

scores = [
    55, 60, 62, 65, 67, 70, 72,
    72, 75, 78, 80, 82, 85, 88,
    90, 92, 95
]

fig, axs = plt.subplots(2, 2)
axs[0, 0].plot(x, y)
axs[0, 0].set_title("Line chart")
axs[0, 0].set_xlabel("X")
axs[0, 0].set_ylabel("Y")

axs[0, 1].bar(categories, values)
axs[0, 1].set_title("Bar char")
axs[0, 1].set_xlabel("Categories")
axs[0, 1].set_ylabel("Values")

axs[1, 0].scatter(x_scatter, y_scatter)
axs[1, 0].set_title("Scatter plot")
axs[1, 0].set_xlabel("X")
axs[1, 0].set_ylabel("Y")

axs[1, 1].hist(scores, bins=8)
axs[1, 1].set_title("Histogram")
axs[1, 1].set_xlabel("Score")
axs[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
````

### Exercise 13 — Annotate a Maximum

Given:

```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 40, 30]
```

Create a line chart and annotate the point with the highest value.

The annotation should indicate:

```text
Maximum
```

and point to the corresponding data point.


````python
def find_location_max(x:list, y:list)->tuple:
    max_y = max(y)
    idx = y.index(max_y)
    return (x[idx], max_y)
    
x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 40, 30]
xy_max = find_location_max(x, y)
xy_text_max = (xy_max[0], xy_max[1] + 2)
plt.plot(x, y, marker="o", linestyle="--", linewidth=2)
plt.title("Annotate a maximum")
plt.xlabel("X")
plt.ylabel("Y")
plt.annotate("Maximum", xy=xy_max, xytext=xy_text_max, arrowprops={"arrowstyle": "->"})
plt.show()
````

### Exercise 17 — Correlation Visualization

Generate two numerical datasets:

```python
x = np.arange(1, 101)
y = 2 * x + np.random.normal(0, 20, 100)
```

Create a scatter plot.

Add a line representing the general trend of the data.

````python
x = np.arange(1, 101)
y = 2 * x + np.random.normal(0, 20, 100)
slope, intercept = np.polyfit(x, y, 1)
trend = slope * x + intercept
print(f"trend = {slope}*x + {intercept}")

plt.scatter(x, y, alpha=0.7, label="Data")

# Trend line
plt.plot(x, trend, linewidth=2, label="Trend line")

plt.legend()
plt.title("Correlation Visualization")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
````

### Exercise 18 — Create a Dashboard

Create a figure containing four visualizations for a fictional company:

```text
1. Monthly Revenue      → Line Chart
2. Product Sales        → Bar Chart
3. Customer Ages        → Histogram
4. Revenue vs Customers → Scatter Plot
```

Use `plt.subplots()`.

Make sure the four charts have:

* Appropriate titles
* Axis labels
* Good spacing
* Readable tick labels

````python
# 1. Monthly Revenue
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [
    12000, 15000, 14000, 18000,
    21000, 19500, 23000, 25000,
    27000, 26000, 30000, 32000
]

fig, axs = plt.subplots(2, 2)
axs[0, 0].plot(months, revenue, marker="o", linestyle="--", linewidth=2)
axs[0, 0].set_title("Monthly Revenue")
axs[0, 0].set_xlabel("month")
axs[0, 0].set_ylabel("revenue")
axs[0, 0].tick_params(axis="x", rotation=45)
axs[0, 0].grid(True, alpha=0.3)

# 2. Product sales
products = ["Laptop", "Phone", "Tablet", "Headphone"]
sales = [120, 200, 150, 180]
axs[0, 1].bar(products, sales)
axs[0, 1].set_title("Product sales")
axs[0, 1].set_xlabel("product")
axs[0, 1].set_ylabel("sales")
axs[0, 1].tick_params(axis="x", rotation=20)
axs[0, 1].grid(True, alpha=0.2)

# 3. Customer Ages
np.random.seed(42)
ages = np.random.normal(35, 10, 200)
ages = np.clip(ages, 18, 70)
axs[1, 0].hist(ages, bins=8)
axs[1, 0].hist(
    ages,
    bins=10,
    edgecolor="black"
)
axs[1, 0].set_title("Customer Ages")
axs[1, 0].set_xlabel("Age")
axs[1, 0].set_ylabel("Number of Customers")

# 4. Revenue vs Customers
customers = np.array([
    100, 120, 150, 170, 200,
    220, 250, 280, 300, 330
])

customer_revenue = np.array([
    12000, 14500, 17000, 19000, 22000,
    24000, 27000, 30000, 32500, 35000
])

axs[1, 1].scatter(
    customers,
    customer_revenue,
    s=60
)
axs[1, 1].set_title("Revenue vs Customers")
axs[1, 1].set_xlabel("Number of Customers")
axs[1, 1].set_ylabel("Revenue ($)")
axs[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### Exercise 20 — Mini Data Visualization Project

Create a fictional student dataset:

```python
students = [
    "Alice", "Bob", "Charlie",
    "David", "Emma", "Frank",
    "Grace", "Henry"
]

math = [85, 70, 92, 65, 88, 76, 95, 80]
english = [90, 75, 89, 70, 94, 78, 91, 82]
science = [88, 72, 95, 68, 90, 80, 93, 85]
```

Build a mini visualization report containing at least **4 charts**:

1. Bar chart showing each student's average score.
2. Line chart comparing Math, English, and Science.
3. Histogram showing the distribution of all scores.
4. Scatter plot showing the relationship between Math and Science.

Then calculate:

* Average score for each student
* Average score for each subject
* Highest individual score
* Lowest individual score
* Student with the highest average

Finally, save the visualization as an image:

```python
plt.savefig("student_analysis.png", dpi=300)
```

````python
students = [
    "Alice", "Bob", "Charlie", "David",
    "Emma", "Frank", "Grace", "Henry"
]

math = [85, 70, 92, 65, 88, 76, 95, 80]
english = [90, 75, 89, 70, 94, 78, 91, 82]
science = [88, 72, 95, 68, 90, 80, 93, 85]

# 1. Create DataFrame
df = pd.DataFrame({
    "Student": students,
    "Math": math,
    "English": english,
    "Science": science
})

subjects = ["Math", "English", "Science"]

# 2. Calculate average score for each student
df["Average"] = df[subjects].mean(axis=1).round(2)

# 3. Calculate average score for each subject
subject_averages = df[subjects].mean().round(2)

# 4. Find highest and lowest individual scores
all_scores = df[subjects].to_numpy().flatten()

highest_score = all_scores.max()
lowest_score = all_scores.min()

# 5. Find student with the highest average
top_student = df.loc[df["Average"].idxmax()]

# 6. Create four charts
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1: Average score by student
axs[0, 0].bar(df["Student"], df["Average"])
axs[0, 0].set_title("Average Score by Student")
axs[0, 0].set_xlabel("Student")
axs[0, 0].set_ylabel("Average Score")
axs[0, 0].tick_params(axis="x", labelrotation=25)

# Chart 2: Compare three subjects
for subject in subjects:
    axs[0, 1].plot(
        df["Student"],
        df[subject],
        marker="o",
        label=subject
    )

axs[0, 1].set_title("Math vs English vs Science")
axs[0, 1].set_xlabel("Student")
axs[0, 1].set_ylabel("Score")
axs[0, 1].tick_params(axis="x", labelrotation=25)
axs[0, 1].legend()

# Chart 3: Histogram of all scores
axs[1, 0].hist(all_scores, bins=8, edgecolor="black")
axs[1, 0].set_title("Distribution of All Scores")
axs[1, 0].set_xlabel("Score")
axs[1, 0].set_ylabel("Frequency")

# Chart 4: Math vs Science
axs[1, 1].scatter(df["Math"], df["Science"])
axs[1, 1].set_title("Math vs Science")
axs[1, 1].set_xlabel("Math Score")
axs[1, 1].set_ylabel("Science Score")

plt.tight_layout()

# Save the visualization
plt.savefig("student_analysis.png", dpi=300, bbox_inches="tight")
plt.show()

# 7. Print analysis results
print("Average score for each student:")
print(df[["Student", "Average"]].to_string(index=False))

print("\nAverage score for each subject:")
print(subject_averages)

print("\nHighest individual score:", highest_score)
print("Lowest individual score:", lowest_score)

print(
    f"Student with the highest average: "
    f"{top_student['Student']} ({top_student['Average']:.2f})"
)
````
