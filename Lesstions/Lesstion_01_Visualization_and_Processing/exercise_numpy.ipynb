### Exercise 19 — Normalize Data

Given:

```python
data = np.array([10, 20, 30, 40, 50])
```

Perform **Min-Max normalization** using:

```text
x_normalized = (x - min(x)) / (max(x) - min(x))
```

The resulting values should be between `0` and `1`.

````python
def normalization(x: np.int64, min_value: np.int64, max_value: np.int64) -> np.float64:
    return (x - min_value)/(max_value - min_value)
data = np.array([10, 20, 30, 40, 50])
min_value = np.min(data)
max_value = np.max(data)
result = [normalization(x, min_value, max_value) for x in data]
print(result)
````

### Exercise 20 — Mini Data-Science Task

Given:

```python
students = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [95, 92, 88],
    [60, 68, 70],
    [85, 80, 84]
])
```

Each row represents a student and each column represents a subject.

Perform the following tasks:

1. Calculate the average score for each student.
2. Calculate the average score for each subject.
3. Find the student with the highest average.
4. Find the student with the lowest average.
5. Find the highest score in the entire dataset.
6. Find the lowest score in the entire dataset.
7. Count how many scores are greater than or equal to `80`.
8. Convert all scores to normalized values between `0` and `1`.

A possible solution should make extensive use of NumPy operations rather than Python `for` loops.

````python
def avg_score_each_student(students: np.array)->list:
    results = []
    rows, _ = students.shape
    for row in range(rows):
        avg = np.round(np.mean(students[row, :]), 2)
        results.append(avg)
    return results

def avg_score_each_subject(students: np.array)->list:
    results = []
    _, cols = students.shape
    for col in range(cols):
        avg = np.round(np.mean(students[:, col]), 2)
        results.append(avg)
    return results
    
def max_avg_score_student(students: np.array) -> tuple:
    avg_scores = np.round(np.mean(students, axis=1), 2)
    idx = np.argmax(avg_scores)
    return (idx, avg_scores[idx])

def min_avg_score_student(students: np.array) -> tuple:
    avg_scores = np.round(np.mean(students, axis=1), 2)
    idx = np.argmin(avg_scores)
    return (idx, avg_scores[idx])

def max_score_dataset(students: np.array) -> tuple:
    row, col = np.unravel_index(np.argmax(students), students.shape)
    return ((row, col), students[row, col])

def min_score_dataset(students: np.array) -> tuple:
    row, col = np.unravel_index(np.argmin(students), students.shape)
    return ((row, col), students[row, col])

def count_score_greater_k(students: np.array, k: int) -> int:
    return len(students[students >= 80])

def normalization_students(students: np.array) -> np.array:
    min_value = np.min(students)
    max_value = np.max(students)

    return np.round(
        (students - min_value) / (max_value - min_value),
        2
    )
    
def main():
    students = np.array([
        [80, 75, 90],
        [65, 70, 72],
        [95, 92, 88],
        [60, 68, 70],
        [85, 80, 84]
    ])
    print(f"Calculate the average score for each student: {avg_score_each_student(students)}")
    print(f"Calculate the average score for each subject: {avg_score_each_subject(students)}")
    print(f"Find the student with highest average: {max_avg_score_student(students)}")
    print(f"Find the student with the lowest average: {min_avg_score_student(students)}")
    print(f"Find the highest score in the entire dataset: {max_score_dataset(students)}")
    print(f"Find the lowest score in the entire dataset {min_score_dataset(students)}")
    print(f"Count how many scores are greater than or equal to 80: {count_score_greater_k(students, k=80)}")
    print(f"Convert all scores to normalized values between 0 and 1: {nomalization_students(students)}")
if __name__ == "__main__":
    main()
````
