### Exercise 4 — Filter Data

Find students who:

1. Have a score greater than 85.

3. Have a score between 80 and 90.
4. Are at least 21 years old and have a score above 85.

---

### Exercise 6 — Handle Missing Values

Given:

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, None, 30, None],
    "salary": [3000, 4000, None, 3500],
    "city": ["Hanoi", "HCMC", None, "Danang"]
})
```

Tasks:

1. Count missing values.
2. Fill missing ages with the median.
3. Fill missing salaries with the mean.
4. Fill missing cities with `"Unknown"`.
5. Verify that the missing values have been handled.

````python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, None, 30, None],
    "salary": [3000, 4000, None, 3500],
    "city": ["Hanoi", "HCMC", None, "Danang"]
})
print("Count missing values")
print(df.isna().sum())
df["age"] = df["age"].fillna(df["age"].mean())
df["salary"] = df["salary"].fillna(df["salary"].mean())
df["city"] = df["city"].fillna("Unknown")
print(df)
````

### Exercise 7 — Sort and Analyze

Given an employee DataFrame:

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "salary": [3000, 4200, 2800, 5000, 3500],
    "age": [25, 30, 28, 35, 27]
})
```

Tasks:

1. Sort by salary ascending.
2. Sort by salary descending.
3. Find the employee with the highest salary.
4. Find the average salary.
5. Find the median salary.

---

````python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "salary": [3000, 4200, 2800, 5000, 3500],
    "age": [25, 30, 28, 35, 27]
})
print("Sort by salary ascending!")
print(df.sort_values(by="salary", ascending=True))
print("Sort by salary descending")
print(df.sort_values(by="salary", ascending=False))
print("Employee with the higest salary:")
print(df.loc[df["salary"].idxmax()])
print(f"Average salary = {df['salary'].mean()}")
print(f"Median salary = {df['salary'].median()}")
````

### Exercise 8 — GroupBy

Given:

```python
df = pd.DataFrame({
    "employee": ["A", "B", "C", "D", "E", "F"],
    "department": [
        "IT", "IT", "HR", "HR", "Finance", "Finance"
    ],
    "salary": [3000, 3500, 2800, 3200, 4000, 4500]
})
```

Calculate:

1. Average salary by department.
2. Maximum salary by department.
3. Minimum salary by department.
4. Employee count by department.
5. Total salary by department.

---

````python
df = pd.DataFrame({
    "employee": ["A", "B", "C", "D", "E", "F"],
    "department": [
        "IT", "IT", "HR", "HR", "Finance", "Finance"
    ],
    "salary": [3000, 3500, 2800, 3200, 4000, 4500]
})
print("Average salary by department")
print(df.groupby("department")["salary"].mean())
print("Maximun salary by department")
print(df.groupby("department")["salary"].max())
print("Minimun salary by department")
print(df.groupby("department")["salary"].min())
print("Total salry by department")
print(df.groupby("department")["salary"].agg(["sum"]))
````

### Exercise 9 — Merge DataFrames

Given:

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"]
})

salary = pd.DataFrame({
    "employee_id": [1, 2, 3, 5],
    "salary": [3000, 3500, 4000, 4500]
})
```

Perform:

1. Inner join
2. Left join
3. Right join
4. Outer join

Then explain the difference between the results.

````python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"]
})

salary = pd.DataFrame({
    "employee_id": [1, 2, 3, 5],
    "salary": [3000, 3500, 4000, 4500]
})

inner_join = pd.merge(employees, salary, on="employee_id", how="inner")
left_join = pd.merge(employees, salary, on="employee_id", how="left")
right_join = pd.merge(employees, salary, on="employee_id", how="right")
outer_join = pd.merge(employees, salary, on="employee_id", how="outer")
print("Inner Join:")
print(inner_join)
print("Left Join:")
print(left_join)
print("Right Join:")
print(right_join)
print("Outer Join:")
print(outer_join)
````

### Exercise 10 — String Processing

Given:

```python
df = pd.DataFrame({
    "name": [" Alice ", "BOB", " Charlie", "david ", " EMMA "],
    "email": [
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@gmail.com",
        "david@gmail.com",
        "emma@yahoo.com"
    ]
})
```

Tasks:

1. Remove whitespace from names.
2. Convert names to lowercase.
3. Create uppercase names.
4. Extract the email domain.
5. Find users with Gmail addresses.

---

````python
df = pd.DataFrame({
    "name": [" Alice ", "BOB", " Charlie", "david ", " EMMA "],
    "email": [
        "alice@gmail.com",
        "bob@yahoo.com",
        "charlie@gmail.com",
        "david@gmail.com",
        "emma@yahoo.com"
    ]
})

df["name"] = df["name"].str.strip()
df["name"] = df["name"].str.lower()
df["name"] = df["name"].str.upper()
# Extract domain from email
df["domain"] = df["email"].str.split("@").str[1]
print(df)
email = "emma@yahoo.com"
if df.loc[df["email"] == email].empty:
    print(f"{email} not found in the DataFrame")
else:
    print(f"{email} found in the DataFrame")
````

### Exercise 11 — Date Processing

Given:

```python
df = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5],
    "order_date": [
        "2026-01-15",
        "2026-02-20",
        "2026-03-10",
        "2026-03-25",
        "2026-04-05"
    ],
    "amount": [100, 250, 180, 300, 220]
})
```

Tasks:

1. Convert `order_date` to datetime.
2. Extract year, month, and day.
3. Find orders in March.
4. Calculate total sales.

````python
df = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5],
    "order_date": [
        "2026-01-15",
        "2026-02-20",
        "2026-03-10",
        "2026-03-25",
        "2026-04-05"
    ],
    "amount": [100, 250, 180, 300, 220]
})
df["order_date"] = pd.to_datetime(df["order_date"])
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["day"] = df["order_date"].dt.day
print(df)
find_march = df[df["order_date"].dt.month == 3]
if find_march.empty:
    print("No orders found in March")
else:
    print("Orders found in March:")
    print(find_march)
total_amount = pd.DataFrame()
total_amount["month"] = df["order_date"].dt.month.unique()
total_amount["amount"] = df.groupby(df["order_date"].dt.month)["amount"].sum().values
print("Total amount by month:")
print(total_amount)
````
